# Примеры использования PDE Solver API

## 1. Уравнение теплопроводности
Пример вызова для 1D случая:

```python
from pde_solvers_api.solvers.heat_equation import solve_heat_equation
import numpy as np

# Параметры
alpha = 0.01
length = 1.0
nx = 100
nt = 500
dt = 0.001
u0 = np.sin(np.linspace(0, np.pi, nx))
bc = {"left": 0.0, "right": 0.0}

solution = solve_heat_equation(alpha, length, nx, nt, dt, u0, bc)
```

## 2. Уравнение Шрёдингера (2D)
Пример для круговой полости:

```python
from pde_solvers_api.solvers.schrodinger_2d import solve_2d_schrodinger
import matplotlib.pyplot as plt

# Параметры
R = 5.0  # радиус полости
N = 50   # точек на измерение

# Решение
E, psi, X, Y = solve_2d_schrodinger(R, N)

print("Первые 5 уровней энергии:", E[:5])

# Визуализация основного состояния
plt.figure(figsize=(8, 6))
plt.contourf(X, Y, psi[0], levels=20, cmap='viridis')
plt.colorbar()
plt.title(f'2D волновая функция (E={E[0]:.3f})')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
